```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("game_data.csv")

df.head()
```

```python

# Dataset info
df.info()

# Statistical summary
df.describe()
```

```python
# Convert timestamp to datetime
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Line plot: Max Speed over time
plt.figure()
plt.plot(df["timestamp"], df["max_speed"])
plt.xlabel("Time")
plt.ylabel("Max Speed")
plt.title("Max Speed Over Time")
plt.show()
```
